# Time-based analysis

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['MonthName'] = df['Date'].dt.month_name()
```

```python
monthly_sales = df.groupby('MonthName')['TotalPrice'].sum()
print("Overall monthly sales is :\n", monthly_sales)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales)
plt.title("Monthly revenue")
plt.xlabel("Monthly_sales")
plt.ylabel("TotalPrice")
plt.xticks(rotation=45)
plt.show()
```

# Customer analysis

```python
customer_analysis = (
    df.groupby('CustomerID')
      .agg(
          Orders=('OrderID', 'nunique'),
          Total_Spent=('TotalPrice', 'sum'),
          Average_Order_Value=('TotalPrice', 'mean')
      )
      .sort_values('Total_Spent', ascending=False)
)

print("customer_analysis :\n\n" ,customer_analysis.head(10))
```

```python
top10 = customer_analysis.head(10).reset_index()
top10['CustomerID'] = top10['CustomerID'].astype(str)

plt.figure(figsize=(10, 6))
sns.barplot(x='Total_Spent', y='CustomerID', data=top10, orient='h')
plt.title("Top 10 Customers by Spending")
plt.xlabel("Total Spending")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()
```

# KEY INSIGHTS

### Key Insights

1. The dataset contains **1,200 orders** from **1,189 unique customers**, generating total revenue of approximately **1.26 million** with an average order value of **1,053.97**.

2. Customer retention is very low. Only **11 customers (0.93%)** placed more than one order, while **99.07%** of customers placed only one order.

3. **Chair** generated the highest product revenue at **195,620.11**, followed very closely by **Printer** at **195,612.61**.

4. **Credit Card** generated the highest revenue among payment methods, contributing approximately **263,847.63**.

5. **June** recorded the highest monthly revenue at **170,616.13**, while **September** recorded the lowest at **69,321.65**.

6. **FREESHIP** was the most frequently used coupon and generated the highest coupon-associated revenue of **335,036.99**.

7. The correlation between **UnitPrice and Quantity is 0.015**, indicating almost no linear relationship between product price and quantity ordered in this dataset.

8. **Cancelled orders** were associated with the highest TotalPrice among order statuses at **276,396.21**. However, this amount should not automatically be considered realized revenue because cancelled orders may represent lost sales.

9. **Instagram** was the most common referral source by order count, generating **259 orders**.


**BUSINESS RECOMMENDATIONS**

### Business Recommendations

1. **Improve customer retention**
   The repeat customer rate is only 0.93%. The business should consider loyalty programs, personalized offers, email campaigns, and post-purchase engagement to encourage repeat purchases.

2. **Focus on high-performing products**
   Chair and Printer generated the highest product revenue. The business should maintain sufficient inventory and consider targeted promotions for these products.

3. **Investigate cancelled orders**
   Cancelled orders are associated with a large amount of TotalPrice. The business should investigate the main reasons for cancellations and identify opportunities to reduce lost sales.

4. **Evaluate promotional strategies**
   FREESHIP has the highest coupon usage and coupon-associated revenue. The business should evaluate whether this promotion improves customer acquisition and order value while remaining financially beneficial.

5. **Investigate seasonal performance**
   June generated the highest monthly revenue, while September generated the lowest. The business should investigate the factors behind these differences and consider seasonal marketing strategies.

6. **Strengthen high-performing referral channels**
   Instagram generated the highest number of orders among referral sources. The business could investigate why Instagram performs well and consider allocating more marketing attention to successful channels.

7. **Use customer data for retention strategies**
   Since most customers purchased only once, customer segmentation and targeted re-engagement campaigns could help increase repeat purchases and customer lifetime value.
