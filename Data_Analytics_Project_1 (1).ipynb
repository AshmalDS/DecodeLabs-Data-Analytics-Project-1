**Handling missing values and checking missing values after cleaning**

```python
print("Filling missing CouponCode with No_CouponCode")
df['CouponCode'] = df['CouponCode'].fillna('No_CouponCode')

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
df.isnull().sum().sum()
```

```python
print("Invalid dates found:", df['Date'].isna().sum())
```

**Text cleaning**

```python
df["Product"]=df["Product"].str.strip()
```

**Final validation**

```python
print("FINAL DATA VALIDATION")
print("----------------------")

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("Duplicate OrderIDs:", df["OrderID"].duplicated().sum())

print("Invalid dates:", df["Date"].isna().sum())

print("\nData types:")
print(df.dtypes)
```
